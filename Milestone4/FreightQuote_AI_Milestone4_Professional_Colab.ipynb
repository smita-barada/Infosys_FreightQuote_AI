## 🌐 Launching Streamlit

The next cell starts the application on port **8501**.

For a public URL, add `NGROK_AUTHTOKEN` to Colab Secrets. If no token is available, the app still starts locally and the notebook prints the local endpoint.


```python
# 7️⃣ Start FreightQuote AI
import subprocess, time, os, signal

os.chdir('/content/freight_app')

# Stop an older Streamlit process if one exists
subprocess.run("pkill -f 'streamlit run' || true", shell=True)

log_file = open('/content/freightquote_streamlit.log', 'w')

process = subprocess.Popen(
    ['streamlit', 'run', 'app.py',
     '--server.port=8501',
     '--server.address=0.0.0.0',
     '--server.headless=true',
     '--browser.gatherUsageStats=false'],
    stdout=log_file,
    stderr=subprocess.STDOUT
)

time.sleep(8)

if process.poll() is None:
    print('✅ Streamlit is running on port 8501')
else:
    print('❌ Streamlit stopped. Showing log:')
    print(open('/content/freightquote_streamlit.log').read()[-5000:])
```

```python
# 8️⃣ Create a public HTTPS URL using ngrok
# Add NGROK_AUTHTOKEN to Colab Secrets for this cell to create a public URL.

from pyngrok import ngrok
import os

try:
    from google.colab import userdata
    ngrok_token = userdata.get('NGROK_AUTHTOKEN')
except Exception:
    ngrok_token = os.getenv('NGROK_AUTHTOKEN')

if ngrok_token:
    ngrok.set_auth_token(ngrok_token)
    tunnel = ngrok.connect(8501, bind_tls=True)
    print('🚀 Public FreightQuote AI URL:')
    print(tunnel.public_url)
else:
    print('ℹ️ NGROK_AUTHTOKEN is not configured.')
    print('Add NGROK_AUTHTOKEN to Colab Secrets and rerun this cell for a public URL.')
    print('Local endpoint: http://127.0.0.1:8501')
```
